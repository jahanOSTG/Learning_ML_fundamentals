```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
newsgroup=fetch_20newsgroups(subset='train',categories=['sci.space'])
```

```python
texts=newsgroup.data[:100]
```

```python
texts
```

```python
vectorizer=CountVectorizer(stop_words='english',max_features=100)
x=vectorizer.fit_transform(texts)
```

```python
print(x)
```

```python
word_freq=dict(zip(vectorizer.get_feature_names_out(),x.toarray().sum(axis=0)))
```

```python
word_freq
```

```python
wordcloud=WordCloud(width=800,height=800,background_color='white').generate_from_frequencies(word_freq)
plt.figure(figsize=(12,6))
plt.imshow(wordcloud,interpolation='hamming')
plt.axis('off')
plt.tight_layout('Top words in Sci.Space article')
plt.show()
```

**Multidimentional Data visualization**

```python
from sklearn.datasets import load_wine
from sklearn.manifold import TSNE
import seaborn as sns
import pandas as pd
```

```python
data=load_wine()
data
```

```python
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df.head()
```

```python
df.shape
```

```python
tsne=TSNE(n_components=2,random_state=42,perplexity=30)
tsne_results=tsne.fit_transform(data.data)
```

```python
tsne_results
```

```python
df['tsne-sd-one']=tsne_results[:,0]
df['tsne-sd-two']=tsne_results[:,1]
display(df.head())
```

```python
plt.figure(figsize=(10,7))
sns.scatterplot(data=df,x='tsne-sd-one',y='tsne-sd-two',hue='target',palette='Set1')
plt.title('t-sne on wine dataset')
plt.show()
```

**Comparetive data visualization**

```python
df=sns.load_dataset('titanic')
```

```python
df.head()
```

```python
plt.figure(figsize=(10,7))
sns.countplot(data=df,x='class',hue='sex')
plt.title('Passerger class distribution by gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,7))
df['survived_label'] = df['survived'].map({0: 'no', 1: 'yes'})
sns.boxplot(x='survived_label',y='age',data=df)
plt.title('Age distribution: Survived vs Not Survived')
plt.xlabel('Survived')
plt.show()
```
